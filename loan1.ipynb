```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Loan_Default.csv.zip")

print("========== DATASET INFORMATION ==========")

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all"))


print("\n========== DEFAULT DISTRIBUTION ==========")

print(df["Status"].value_counts())

print("\nDefault Percentage:")

print(
    df["Status"]
    .value_counts(normalize=True)
    .mul(100)
)


# Default distribution

plt.figure(figsize=(7,5))

df["Status"].value_counts().plot(
    kind="bar"
)

plt.title("Loan Default Distribution")

plt.xlabel("Status")

plt.ylabel("Number of Customers")

plt.show()


# Credit score

plt.figure(figsize=(8,5))

df.groupby("Status")["Credit_Score"].mean().plot(
    kind="bar"
)

plt.title("Average Credit Score by Default")

plt.xlabel("Status")

plt.ylabel("Average Credit Score")

plt.show()


# Income

plt.figure(figsize=(8,5))

df.groupby("Status")["income"].mean().plot(
    kind="bar"
)

plt.title("Average Income by Default")

plt.xlabel("Status")

plt.ylabel("Average Income")

plt.show()


# Loan amount

plt.figure(figsize=(8,5))

df.groupby("Status")["loan_amount"].mean().plot(
    kind="bar"
)

plt.title("Average Loan Amount by Default")

plt.xlabel("Status")

plt.ylabel("Average Loan Amount")

plt.show()


# Loan purpose

plt.figure(figsize=(10,5))

pd.crosstab(
    df["loan_purpose"],
    df["Status"]
).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Loan Purpose vs Default")

plt.xlabel("Loan Purpose")

plt.ylabel("Number of Customers")

plt.xticks(rotation=45)

plt.show()
```
